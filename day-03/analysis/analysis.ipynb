# Facility Data Analysis

## Objective
Analyze the cleaned facility dataset to calculate key statistics and identify useful insights about cleanliness, odor, waste levels, water availability, footfall, and complaints.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../dataset/cleaned_facility_data.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python

print("Summary Statistics:")
df[["cleanliness_score", "odor_score", "footfall", "complaints"]].describe().round(2)
```

```python

print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nDataset Shape:", df.shape)
```

```python
print("Waste Level Counts:")
print(df["waste_level"].value_counts())

print("\nWater Availability:")
print(df["water_availability"].value_counts())

print("\nAverage Cleanliness by Location:")
print(
    df.groupby("location")["cleanliness_score"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)
```

### Key Findings

1. **Water availability:** 26% of the facilities report unavailable water, indicating a need to investigate water access.
2. **Waste management:** 33% of facilities fall into the high-waste category, suggesting that waste management needs attention at these facilities.
3. **Cleanliness variation:** Residential facilities have the highest average cleanliness score (9.00), whereas Industrial facilities have the lowest (5.43).
4. **Locations requiring attention:** Industrial, Market, and East have relatively low average cleanliness scores and may benefit from further inspection.
5. **Potential outliers:** One facility has footfall of 5,000 and another has 100 complaints. These observations should be verified before drawing conclusions.
