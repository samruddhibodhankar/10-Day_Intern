# Data Cleaning

## Objective
Inspect the facility dataset and identify missing values, duplicate records, invalid values, and potential outliers before preparing the data for analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../dataset/facility_data.csv")

df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values per column:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("Duplicate facility IDs:", df["facility_id"].duplicated().sum())
```

```python
print("Cleanliness scores outside 1–10:")
print(df[~df["cleanliness_score"].between(1, 10)])

print("\nOdor scores outside 1–10:")
print(df[~df["odor_score"].between(1, 10)])

print("\nUnexpected waste levels:")
print(df[~df["waste_level"].isin(["Low", "Medium", "High"])])

print("\nUnexpected water availability values:")
print(df[~df["water_availability"].isin(["Yes", "No"])])
```

```python
print("Footfall summary:")
print(df["footfall"].describe())

print("\nComplaints summary:")
print(df["complaints"].describe())
```

```python
print("Potential outliers in footfall:")
print(df[df["footfall"] > 1000][["facility_id", "footfall"]])

print("\nPotential outliers in complaints:")
print(df[df["complaints"] > 50][["facility_id", "complaints"]])
```

## Data Cleaning

The inspection identified missing cleanliness and odor scores, invalid cleanliness scores outside the 1–10 range, and two potential outliers in footfall and complaints. No duplicate rows or duplicate facility IDs were found.

In the following steps, invalid scores will be replaced with missing values, missing scores will be filled using the median, and inspection dates will be converted to datetime format. Potential outliers will be retained because they may represent genuine observations.

```python
# Create a copy to preserve the original dataset
cleaned_df = df.copy()

# Replace invalid cleanliness scores with missing values
invalid_cleanliness = (
    cleaned_df["cleanliness_score"].notna()
    & ~cleaned_df["cleanliness_score"].between(1, 10)
)

cleaned_df.loc[invalid_cleanliness, "cleanliness_score"] = np.nan

print("Invalid cleanliness scores corrected:",
      invalid_cleanliness.sum())

print("\nRemaining missing values:")
print(cleaned_df[["cleanliness_score", "odor_score"]].isnull().sum())
```

```python
# Fill missing scores using the median of each column
cleaned_df["cleanliness_score"] = cleaned_df["cleanliness_score"].fillna(
    cleaned_df["cleanliness_score"].median()
)

cleaned_df["odor_score"] = cleaned_df["odor_score"].fillna(
    cleaned_df["odor_score"].median()
)

print("Missing values after cleaning:")
print(cleaned_df[["cleanliness_score", "odor_score"]].isnull().sum())
```

```python
# Convert inspection_date to datetime format
cleaned_df["inspection_date"] = pd.to_datetime(
    cleaned_df["inspection_date"]
)

print("Inspection date data type:",
      cleaned_df["inspection_date"].dtype)

print("\nDate range:")
print("First inspection:", cleaned_df["inspection_date"].min())
print("Last inspection:", cleaned_df["inspection_date"].max())
```

```python
print("Dataset shape:", cleaned_df.shape)

print("\nMissing values:")
print(cleaned_df.isnull().sum())

print("\nDuplicate rows:", cleaned_df.duplicated().sum())

print("\nCleanliness score range:",
      cleaned_df["cleanliness_score"].min(), "to",
      cleaned_df["cleanliness_score"].max())

print("Odor score range:",
      cleaned_df["odor_score"].min(), "to",
      cleaned_df["odor_score"].max())
```

```python
cleaned_df.to_csv("../dataset/cleaned_facility_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
