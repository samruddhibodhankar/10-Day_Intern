# Facility Data Visualizations

## Objective
This notebook creates visualizations to understand facility cleanliness, waste levels, footfall, and complaints using the cleaned facility dataset.

## Dataset
The notebook uses `cleaned_facility_data.csv`, which contains 100 facility records and 9 columns.

## Visualizations
The following charts will be created:

1. Bar chart — Average Cleanliness Score by Location
2. Bar chart — Number of Facilities by Waste Level
3. Histogram — Distribution of Facility Footfall
4. Scatter plot — Footfall vs. Complaints
5. Box plot — Distribution of Complaints

The charts will be saved as PNG files in the `visualizations` folder.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../dataset/cleaned_facility_data.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python

average_cleanliness = df.groupby("location")["cleanliness_score"].mean()

average_cleanliness.plot(kind="bar", figsize=(10, 5))
plt.title("Average Cleanliness Score by Location")
plt.xlabel("Location")
plt.ylabel("Average Cleanliness Score")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("01_cleanliness_by_location.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

waste_counts = df["waste_level"].value_counts()

waste_counts.plot(kind="bar", figsize=(8, 5))
plt.title("Number of Facilities by Waste Level")
plt.xlabel("Waste Level")
plt.ylabel("Number of Facilities")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("02_facilities_by_waste_level.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["footfall"], bins=10, edgecolor="black")

plt.title("Distribution of Facility Footfall")
plt.xlabel("Footfall")
plt.ylabel("Number of Facilities")
plt.tight_layout()

plt.savefig("03_footfall_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["footfall"], df["complaints"], alpha=0.7)

plt.title("Footfall vs. Complaints")
plt.xlabel("Footfall")
plt.ylabel("Number of Complaints")
plt.tight_layout()

plt.savefig("04_footfall_vs_complaints.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

plt.figure(figsize=(8, 5))
plt.boxplot(df["complaints"])

plt.title("Distribution of Complaints Across Facilities")
plt.ylabel("Number of Complaints")
plt.tight_layout()

plt.savefig("05_complaints_boxplot.png", dpi=300, bbox_inches="tight")
plt.show()
```
